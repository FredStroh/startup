# File Selector Widget Example

This notebook demonstrates how to use the file selector widget for Jupyter notebooks.

The file selector allows you to browse and select files from your file system directly within a Jupyter notebook.

## Installation

First, make sure you have the required dependencies installed:

```bash
pip install ipywidgets jupyter
```

## Basic Usage

Import the file selector and create an instance:

````python
from file_selector import create_file_selector

# Create a file selector starting in the current directory
selector = create_file_selector()
````

## How to Use

1. **Navigate**: Click on directories (📁) to navigate into them
2. **Go Up**: Use the "Parent Directory" button to go up one level
3. **Select**: Click on a file (📄) and then click "Select File" button
4. **Get Selection**: Use the `get_selected()` method to retrieve the selected file path

## Retrieve Selected File

After selecting a file using the widget above, run this cell to get the file path:

````python
selected_file = selector.get_selected()
if selected_file:
    print(f"Selected file: {selected_file}")
else:
    print("No file selected yet")
````

## Using Selected File

Once you have the file path, you can use it for various operations:

````python
if selected_file:
    # Example: Read the file content
    with open(selected_file, 'r') as f:
        content = f.read()
        print(f"File content (first 500 characters):\n{content[:500]}")
else:
    print("Please select a file first using the widget above")
````

## Starting from a Specific Directory

You can also start the file selector from a specific directory:

````python
import os

# Start from home directory
home_selector = create_file_selector(start_path=os.path.expanduser('~'))
````
